```python
%%writefile src/data.py
"""Data pipeline: 4-class EuroSAT with a stratified 70/15/15 split.

Images are returned as RAW [0,1] tensors (no mean/std normalization), so
attacks can use epsilon in true pixel units. Normalization lives inside
the model (see model.py).
"""
import random
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import datasets, transforms
from sklearn.model_selection import train_test_split

SEED = 42
KEEP = ["AnnualCrop", "Forest", "Residential", "River"]


def set_seed(seed=SEED):
    """Fixing all random number generators so results are repeatable."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)


class Subset4(Dataset):
    """A view on the EuroSAT dataset restricted to a list of indices.

    Each split (train/val/test) gets its own instance with its own
    transform, which avoids the transform-overwrite bug in the original
    report's random_split code.
    """
    def __init__(self, base, idx, transform, remap):
        self.base = base            # the full EuroSAT dataset (PIL images)
        self.idx = idx              # which images belong to this split
        self.transform = transform  # preprocessing for this split
        self.remap = remap          # original label (0-9) -> new label (0-3)

    def __len__(self):
        return len(self.idx)

    def __getitem__(self, k):
        img, t = self.base[self.idx[k]]#PIL image + original label
        return self.transform(img), self.remap[t]


def get_dataloaders(root="data", batch_size=64, num_workers=2):
    """Download EuroSAT, keep 4 classes, split, and return 3 DataLoaders."""
    #64x64 RGB version; downloaded once into `root`
    base = datasets.EuroSAT(root=root, download=True)

    #mapping the kept classes to new labels 0 to 3
    keep_idx = [base.class_to_idx[c] for c in KEEP]
    remap = {old: new for new, old in enumerate(keep_idx)}
    indices = [i for i, t in enumerate(base.targets) if t in remap]
    labels = [remap[base.targets[i]] for i in indices]

    #stratified split: 70% train, then the other 30% halved into val/test
    tr_i, tmp_i, _, tmp_y = train_test_split(
        indices, labels, test_size=0.30, stratify=labels, random_state=SEED)
    va_i, te_i = train_test_split(
        tmp_i, test_size=0.50, stratify=tmp_y, random_state=SEED)

    #transforms: resize to 224 (what ResNet expects), convert to [0,1] tensor
    train_tf = transforms.Compose([transforms.Resize(224),
                                   transforms.RandomHorizontalFlip(),#light augmentation
                                   transforms.ToTensor()])
    eval_tf = transforms.Compose([transforms.Resize(224),
                                  transforms.ToTensor()])#no augmentation for eval

    train_ds = Subset4(base, tr_i, train_tf, remap)
    val_ds = Subset4(base, va_i, eval_tf, remap)
    test_ds = Subset4(base, te_i, eval_tf, remap)

    #shuffle only the training data
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=num_workers)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=num_workers)
    test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=num_workers)
    return train_loader, val_loader, test_loader
```

```python
%%writefile src/model.py
import torch
import torch.nn as nn
import torchvision.models as models

#ImageNet statistics: the pretrained weights wants inputs normalized this way
IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)


class Normalized(nn.Module):
    """Wraps a model so it accepts RAW [0,1] images.

    Normalization happens inside forward(), so attacks can work directly
    in [0,1] pixel space (epsilon = true pixel units) while gradients
    still flow through the normalization step.
    """
    def __init__(self, model, mean, std):
        super().__init__()
        self.model = model
        #register_buffer: saved with the model and moved to GPU with it, but not treated as a trainable parameter
        self.register_buffer("mean", torch.tensor(mean).view(1, 3, 1, 1))
        self.register_buffer("std",  torch.tensor(std).view(1, 3, 1, 1))

    def forward(self, x):
        return self.model((x - self.mean) / self.std)


def build_model(num_classes=4):
    #1 load ResNet50 with ImageNet-pretrained weights
    backbone = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)
    #2 replace the 1000-class output layer with a fresh 4-class one
    backbone.fc = nn.Linear(backbone.fc.in_features, num_classes)
    #3 wrap it so inputs are normalized inside the model
    return Normalized(backbone, IMAGENET_MEAN, IMAGENET_STD)
```

```python
%%writefile src/train.py
import torch
import torch.nn as nn


def run_epoch(model, loader, device, criterion, optimizer=None):
    """One pass over a dataloader.
    If an optimizer is given -> training mode (weights update).
    If not -> evaluation mode (no weight updates, no gradients).
    Returns (average loss, accuracy).
    """
    training = optimizer is not None
    model.train() if training else model.eval()

    total_loss, correct, total = 0.0, 0, 0
    #torch.set_grad_enabled(False) during evaluation saves memory and time
    with torch.set_grad_enabled(training):
        for x, y in loader:
            x, y = x.to(device), y.to(device)

            logits = model(x)#raw class scores
            loss = criterion(logits, y)

            if training:
                optimizer.zero_grad()#clear old gradients
                loss.backward()#compute new gradients
                optimizer.step()#update the weights

            #accumulate statistics (weight loss by batch size)
            total_loss += loss.item() * x.size(0)
            correct += (logits.argmax(1) == y).sum().item()
            total += x.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def predict_all(model, loader, device):
    """Returns (true labels, predicted labels) as numpy arrays."""
    model.eval()
    ys, ps = [], []
    for x, y in loader:
        preds = model(x.to(device)).argmax(1).cpu()
        ys.append(y); ps.append(preds)
    return torch.cat(ys).numpy(), torch.cat(ps).numpy()
```

```python
import sys; sys.path.append("src")        #lets the notebook import from src/
import torch, torch.nn as nn, torch.optim as optim
from model import build_model
from train import run_epoch
import os

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = build_model(num_classes=4).to(device)
criterion = nn.CrossEntropyLoss()

#small LR (1e-4): cause fine-tuning pretrained weights
optimizer = optim.Adam(model.parameters(), lr=1e-4)

EPOCHS, PATIENCE = 10, 3
#cosine schedule: LR decays smoothly toward 0 over EPOCHS
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

best_val_acc, bad_epochs = 0.0, 0

# Kui Google Drive'i kausta CKPT_DIR pole defineeritud, loome kohaliku kausta
if "CKPT_DIR" not in globals():
    CKPT_DIR = "checkpoints"
    os.makedirs(CKPT_DIR, exist_ok=True)

ckpt_path = f"{CKPT_DIR}/baseline_resnet50.pth"

for epoch in range(EPOCHS):
    tr_loss, tr_acc = run_epoch(model, train_loader, device, criterion, optimizer)
    va_loss, va_acc = run_epoch(model, val_loader,   device, criterion)#no optimizer
    scheduler.step()#advance the LR schedule once per epoch

    print(f"epoch {epoch+1:2d} | train loss {tr_loss:.4f} acc {tr_acc:.4f} "
          f"| val loss {va_loss:.4f} acc {va_acc:.4f}")

    #early stopping: keep the best model on val data
    if va_acc > best_val_acc:
        best_val_acc, bad_epochs = va_acc, 0
        torch.save(model.state_dict(), ckpt_path)
    else:
        bad_epochs += 1
        if bad_epochs >= PATIENCE:
            print("early stopping"); break
```

```python
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from train import predict_all

model.load_state_dict(torch.load(ckpt_path))   #reloading the best checkpoint
y_true, y_pred = predict_all(model, test_loader, device)

print(classification_report(y_true, y_pred, target_names=KEEP, digits=4))

ConfusionMatrixDisplay.from_predictions(
    y_true, y_pred, display_labels=KEEP, normalize="true", cmap="Blues")
os.makedirs("results", exist_ok=True)
plt.savefig("results/baseline_confusion.png", dpi=150, bbox_inches="tight")
plt.show()
```
